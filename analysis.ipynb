```python
##### CIBC Employee Comms

import pandas as pd
import numpy as np

#Loading the df

path = r"C:\Users\csyriac\Downloads\Employee_Comms_Analytics_Exercise_Dataset.xlsx"
dataset = pd.read_excel(path, sheet_name="Communications Data")

#Clean-up
dataset = dataset.drop_duplicates(subset=["Comm ID"])
dataset = dataset.dropna(subset=["Comm ID"])

#Adding reach, engage_rate and click_rate
dataset["reach_%"] = dataset["Unique Viewers"] / dataset["Target Audience Size"] * 100
dataset["engage_rate_%"] = (dataset["Reactions"] + dataset["Comments"] + dataset["Shares"]) / dataset["Unique Viewers"] * 100
dataset["click_rate_%"] = dataset["Link Clicks"] / dataset["Unique Viewers"] * 100

#Announcements give reach, not engagement
dataset.groupby("Post Type")[["reach_%", "engage_rate_%"]].median().sort_values(by="reach_%", ascending=False)

#benefits enrolment, reach VS action
enrol = dataset[dataset["Comm ID"].isin(["C117", "C124"])]
enrol[["Comm ID", "Title", "Post Type", "Unique Viewers", "Link Clicks", "click_rate_%"]].sort_values(by="click_rate_%", ascending=False)

#Conversations
dataset["views_pr_viewer"] = dataset["Total Views"] / dataset["Unique Viewers"]
dataset.nlargest(5, "Comments")[["Comm ID", "Title", "Comments", "views_pr_viewer"]]
```

```python
### Visualizations

import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

path1 = r"C:\Users\csyriac\Downloads"
plt.rcParams.update({"font.size": 11, "axes.spines.top": False,
                     "axes.spines.right": False})
RED, GREY = "#C8102E", "#5A5A5A"

# C1: Announcements vs standard posts
g = dataset.groupby("Post Type")["reach_%"].median()
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.bar(g.index, g.values, color=[RED, GREY])
ax.set_ylabel("Median reach (% of target audience who viewed)")
ax.set_title("Announcements reach ~3x more of the audience")
for i, v in enumerate(g.values):
    ax.text(i, v + 1.5, f"{v:.0f}%", ha="center", fontweight="bold")
fig.tight_layout()
fig.savefig(path1 + "/reach_by_post_type.png", dpi=200)

#C2: benefits enrolment 
b = dataset[dataset["Comm ID"].isin(["C117", "C124"])].set_index("Comm ID")
labels = ["Announcement\n(Oct 13)", "Reminder post\n(Oct 28)"]
fig, axes = plt.subplots(1, 2, figsize=(9, 4))
axes[0].bar(labels, b.loc[["C117", "C124"], "Unique Viewers"], color=[RED, GREY])
axes[0].set_title("Unique viewers")
axes[0].set_ylabel("Employees")
axes[1].bar(labels, b.loc[["C117", "C124"], "Link Clicks"], color=[RED, GREY])
axes[1].set_title("Link clicks (enrolment action)")
fig.suptitle("Benefits enrolment: the push got the views, not the clicks",
             fontweight="bold")
fig.tight_layout()
fig.savefig(path1 + "/benefits_enrolment.png", dpi=200)

#C3: Conversationz
g2 = dataset.groupby("Post Type")["engagement_rate"].median()
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.bar(g2.index, g2.values, color=[RED, GREY])
ax.set_ylabel("Median engagements per viewer (%)")
ax.set_title("Per viewer, standard posts engage more than announcements")
for i, v in enumerate(g2.values):
    ax.text(i, v + 0.2, f"{v:.1f}%", ha="center", fontweight="bold")
fig.tight_layout()
fig.savefig(path1 + "/engagement_rate_by_type.png", dpi=200)
```
